```python
import statistics
numbers = [10, 20, 30, 40, 50]
```

```python
print("Numbers:", numbers)
```

```python
total = sum(numbers)
average = total / len(numbers)
minimum = min(numbers)
maximum = max(numbers)

print("\n--- Arithmetic Calculations ---")
print("Sum:", total)
print("Average:", average)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
mean_value = statistics.mean(numbers)
median_value = statistics.median(numbers)
std_deviation = statistics.stdev(numbers)

print("\n--- Statistical Calculations ---")
print("Mean:", mean_value)
print("Median:", median_value)
print("Standard Deviation:", round(std_deviation, 2))
```

```python
print("sum() - Calculates total")
print("len() - Counts values")
print("min() - Finds smallest value")
print("max() - Finds largest value")
print("statistics.mean() - Calculates mean")
print("statistics.median() - Calculates median")
print("statistics.stdev() - Calculates sample standard deviation")
```
